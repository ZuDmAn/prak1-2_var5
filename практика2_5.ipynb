## Подготовка данных

Создаём две таблицы:
- `sales` — продажи с пропусками в `Price` и `Region`
- `products` — справочник товаров с категориями и весом

```python
# Создание таблицы продаж
sales = pd.DataFrame({
    'OrderID': [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    'Product': ['Laptop', 'Phone', 'Tablet', 'Laptop', 'Phone',
                'Laptop', 'Tablet', 'Phone', 'Laptop', 'Tablet'],
    'Quantity': [1, 2, 1, 3, 1, 2, 1, 4, 1, 2],
    'Price': [50000, 30000, 20000, 50000, np.nan,
              50000, 20000, 30000, 50000, np.nan],
    'Region': ['Moscow', 'SPb', 'Kazan', 'Moscow', 'SPb',
               'Kazan', np.nan, 'SPb', 'Moscow', 'Kazan']
})

# Создание таблицы товаров
products = pd.DataFrame({
    'Product': ['Laptop', 'Phone', 'Tablet', 'Monitor'],
    'Category': ['Electronics', 'Electronics', 'Electronics', 'Accessories'],
    'Weight_kg': [2.5, 0.2, 0.5, 3.0]
})

print("=" * 60)
print("ТАБЛИЦА ПРОДАЖ (sales)")
print("=" * 60)
print(sales)

print("\n" + "=" * 60)
print("ТАБЛИЦА ТОВАРОВ (products)")
print("=" * 60)
print(products)

print("\n" + "=" * 60)
print("ИНФОРМАЦИЯ О SALES")
print("=" * 60)
sales.info()

print("\n" + "=" * 60)
print("ПРОПУСКИ В SALES")
print("=" * 60)
print(sales.isnull().sum())
```

## Задание 1. Среднее значение Quantity для каждого продукта

Используем `.groupby()` для группировки по продукту и `.mean()` для среднего.

```python
# Задание 1: среднее Quantity по продуктам
avg_quantity = sales.groupby('Product')['Quantity'].mean()

print("=" * 60)
print("ЗАДАНИЕ 1: Среднее Quantity по продуктам")
print("=" * 60)
print(avg_quantity)

print("\nС сортировкой по убыванию:")
print(avg_quantity.sort_values(ascending=False))

print("\nРасширенная статистика:")
print(sales.groupby('Product')['Quantity'].agg(['mean', 'median', 'min', 'max', 'count']))
```

## Задание 2. Продукт с максимальной общей выручкой

Общая выручка = `Quantity * Price`. Группируем по продукту и суммируем.

```python
# Задание 2: продукт с максимальной выручкой
sales['Revenue'] = sales['Quantity'] * sales['Price']

print("=" * 60)
print("ЗАДАНИЕ 2: Продукт с максимальной выручкой")
print("=" * 60)

print("\nDataFrame с Revenue:")
print(sales[['OrderID', 'Product', 'Quantity', 'Price', 'Revenue']])

revenue_by_product = sales.groupby('Product')['Revenue'].sum()
print("\nСуммарная выручка по продуктам:")
print(revenue_by_product)

max_revenue_product = revenue_by_product.idxmax()
max_revenue_value = revenue_by_product.max()

print(f"\n>>> Продукт с максимальной выручкой: {max_revenue_product}")
print(f">>> Значение: {max_revenue_value:,.0f}")
```

## Задание 3. Заполнить пропуски в Region методом forward fill

⚠️ В Pandas 3.x метод `fillna(method='ffill')` удалён. Используем `.ffill()`.

```python
# Задание 3: forward fill для Region
print("=" * 60)
print("ЗАДАНИЕ 3: Заполнение пропусков в Region (forward fill)")
print("=" * 60)

print("\n--- До заполнения ---")
print(sales[['OrderID', 'Product', 'Region']])
print(f"\nПропусков в Region: {sales['Region'].isnull().sum()}")

# Создаём копию и заполняем
sales_ffill = sales.copy()
sales_ffill['Region'] = sales_ffill['Region'].ffill()

print("\n--- После forward fill ---")
print(sales_ffill[['OrderID', 'Product', 'Region']])
print(f"\nПропусков в Region: {sales_ffill['Region'].isnull().sum()}")

# Сравнение
print("\n--- Сравнение ---")
comparison = pd.DataFrame({
    'OrderID': sales['OrderID'],
    'Region_before': sales['Region'],
    'Region_after': sales_ffill['Region']
})
print(comparison)
```

## Задание 4. Сортировка по OrderID в обратном порядке

Используем `.sort_values(ascending=False)`.

```python
# Задание 4: сортировка по OrderID в обратном порядке
sales_sorted = sales.sort_values('OrderID', ascending=False)

print("=" * 60)
print("ЗАДАНИЕ 4: Сортировка по OrderID (обратный порядок)")
print("=" * 60)
print(sales_sorted[['OrderID', 'Product', 'Quantity', 'Price', 'Region']])

print("\nПроверка (первые 5 OrderID):")
print(list(sales_sorted['OrderID'].head()))
```

## Задание 5. Outer join таблиц sales и products

`how='outer'` — сохраняет все строки из обеих таблиц, заполняя отсутствующие значения `NaN`.

```python
# Задание 5: outer join
merged_outer = pd.merge(sales, products, on='Product', how='outer')

print("=" * 60)
print("ЗАДАНИЕ 5: Outer join sales и products")
print("=" * 60)
print(merged_outer)

print("\n--- Информация ---")
print(f"Строк в sales:    {len(sales)}")
print(f"Строк в products: {len(products)}")
print(f"Строк в outer:    {len(merged_outer)}")

print("\n--- Строки только из products ---")
only_in_products = merged_outer[merged_outer['OrderID'].isnull()]
print(only_in_products)
```
