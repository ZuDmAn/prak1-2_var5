## Подготовка данных

Создаём исходный DataFrame с информацией о сотрудниках.

```python
# Создание исходного DataFrame
data = {
    'ID': range(1, 11),
    'Name': ['Alice', 'Bob', 'Charlie', 'David', 'Eva',
             'Frank', 'Grace', 'Helen', 'Ivan', 'Julia'],
    'Age': [23, 35, 29, 40, 28, 45, 32, 27, 38, 31],
    'City': ['Moscow', 'SPb', 'Moscow', 'Kazan', 'SPb',
             'Moscow', 'SPb', 'Kazan', 'Moscow', 'SPb'],
    'Salary': [50000, 70000, 60000, 80000, 55000,
               90000, 65000, 58000, 75000, 62000]
}

df = pd.DataFrame(data)
print(df)
```

## Задание 1. Размер DataFrame

Вывести размер таблицы через атрибут `.shape`.

```python
# Задание 1: размер DataFrame
print("Размер DataFrame:", df.shape)
print("Количество строк:", df.shape[0])
print("Количество столбцов:", df.shape[1])
```

## Задание 2. Первые 5 строк и столбцы Name, Age

Используем `.loc[]` для выборки по меткам.

```python
# Задание 2: первые 5 строк, столбцы Name и Age
result = df.loc[:4, ['Name', 'Age']]
print(result)
```

## Задание 3. Строки, где City равен 'Kazan' или 'SPb'

Используем метод `.isin()` для фильтрации по списку значений.

```python
# Задание 3: City == 'Kazan' или 'SPb'
result = df[df['City'].isin(['Kazan', 'SPb'])]
print(result)
```

## Задание 4. Минимальное значение Salary

Находим минимальную зарплату через `.min()`.

```python
# Задание 4: минимальная зарплата
min_salary = df['Salary'].min()
print("Минимальная зарплата:", min_salary)

print("\nСотрудник с минимальной зарплатой:")
print(df[df['Salary'] == min_salary][['Name', 'City', 'Salary']])
```

## Задание 5. Столбец Bonus = 10% от Salary

Создаём новый столбец на основе существующего.

```python
# Задание 5: Bonus = 10% от Salary
df['Bonus'] = df['Salary'] * 0.1

print("DataFrame с новым столбцом Bonus:")
print(df[['Name', 'Salary', 'Bonus']])
```
